# Alternation index (smoke test)

Same steps as `generate_alternation_index.ipynb`, with a small simulation so the pipeline can be checked quickly.

```python
import sys

sys.path.append('../src')

from Utils.indices import AlternationIndex
```

```python
index_gen = AlternationIndex(
    num_points=16,
    num_rounds=5,
    num_episodes=1,
    max_agents=4,
    max_epsilon=0.1,
    seed=42,
    fancy_2P=True,
)
df = index_gen.simulate_data()
print(df['data_type'].value_counts())
```

```python
index_gen.create_index_sklearn()
index_gen.create_index_mlp()
index_gen.create_index_cnn()
```

```python
row = df[index_gen.measures].head(1)
for priority in ('sklearn', 'mlp'):
    loaded = AlternationIndex.from_file(priority=priority)
    print(priority, loaded.classify(row))

sim_id = index_gen.full_data['id_sim'].iloc[0]
one_sim = index_gen.full_data[index_gen.full_data['id_sim'] == sim_id]
loaded_cnn = AlternationIndex.from_file(priority='cnn')
print('cnn', loaded_cnn.classify(one_sim))
```
